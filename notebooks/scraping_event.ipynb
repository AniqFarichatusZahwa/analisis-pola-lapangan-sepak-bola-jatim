### 4.2 Fungsi scraping pertandingan Liga 1 musim 2023/2024

```python
def scrape_persebaya(url, musim):

    response = requests.get(
        url,
        headers={
            'User-Agent': 'Mozilla/5.0'
        },
        timeout=30
    )

    response.raise_for_status()

    soup = BeautifulSoup(response.text, 'html.parser')

    # Ambil seluruh teks halaman
    text = soup.get_text(" ", strip=True)

    # Pola tanggal Indonesia
    bulan = (
        r'Januari|Februari|Maret|April|Mei|Juni|'
        r'Juli|Agustus|September|Oktober|November|Desember'
    )

    pola = rf'(\d{{1,2}}\s+(?:{bulan})\s+\d{{4}})\s+'
    pola += r'(.+?)\s+FT\s+(\d+:\d+)\s+(.+?)\s+'
    pola += r'(GELORA BUNG TOMO)'

    hasil = re.findall(
        pola,
        text,
        flags=re.IGNORECASE
    )

    data = []

    for tanggal, tim_home, skor, tim_away, venue in hasil:

        data.append({
            'musim': musim,
            'tanggal': tanggal,
            'tim_home': tim_home.strip(),
            'skor': skor,
            'tim_away': tim_away.strip(),
            'nama_venue': 'Gelora Bung Tomo',
            'jenis_olahraga': 'Sepak Bola',
            'level_event': 'Nasional',
            'sumber': url
        })

    return pd.DataFrame(data)
```

```python
df_event_list = []

for musim, url in urls_persebaya.items():

    print("Scraping:", musim)

    df = scrape_persebaya(url, musim)

    print("Jumlah pertandingan GBT:", len(df))

    df_event_list.append(df)

data_event_gbt = pd.concat(
    df_event_list,
    ignore_index=True
)

display(data_event_gbt)
```

### 4.3 Ekstraksi tanggal, tim, skor, dan venue

```python
# Ambil bagian fixtures
fixtures = soup.find("div", id="fixtures")

# Semua tabel pertandingan
tables = fixtures.find_all("table", id="table-pertandingan")

# Container tanggal
containers = fixtures.find_all("div", class_="col-md-12")

# Container pertama adalah wrapper besar, jadi ambil mulai index 1
tanggal_containers = containers[1:]

print("Jumlah pertandingan :", len(tables))
print("Jumlah tanggal      :", len(tanggal_containers))
```

```python
bulan_id = {
    "Januari": 1,
    "Februari": 2,
    "Maret": 3,
    "April": 4,
    "Mei": 5,
    "Juni": 6,
    "Juli": 7,
    "Agustus": 8,
    "September": 9,
    "Oktober": 10,
    "November": 11,
    "Desember": 12
}

def parse_tanggal_id(tanggal):
    """
    Mengubah tanggal seperti:
    '8 Juli 2023'
    menjadi datetime.
    """
    match = re.search(
        r'(\d{1,2})\s+'
        r'(Januari|Februari|Maret|April|Mei|Juni|Juli|Agustus|September|Oktober|November|Desember)'
        r'\s+(\d{4})',
        tanggal,
        re.IGNORECASE
    )

    if not match:
        return pd.NaT

    hari = int(match.group(1))
    bulan = bulan_id[match.group(2).capitalize()]
    tahun = int(match.group(3))

    return pd.Timestamp(year=tahun, month=bulan, day=hari)
```

```python
data_liga = []

for i, (table, container) in enumerate(
    zip(tables, tanggal_containers),
    start=1
):

    # -------------------------
    # TANGGAL
    # -------------------------
    tanggal_text = container.get_text(" ", strip=True)
    tanggal = parse_tanggal_id(tanggal_text)

    # -------------------------
    # ROW PERTANDINGAN
    # -------------------------
    row = table.find("tr", class_="table-row")

    if row is None:
        continue

    # -------------------------
    # TIM
    # -------------------------
    tim = row.select("b.club-fixture-team")

    if len(tim) >= 2:
        tim_kandang = tim[0].get_text(" ", strip=True)
        tim_tandang = tim[1].get_text(" ", strip=True)
    else:
        tim_kandang = None
        tim_tandang = None

    # -------------------------
    # SKOR
    # -------------------------
    score_element = row.select_one(".club-fixture-score")

    if score_element:
        score_text = score_element.get_text(" ", strip=True)

        # Ambil skor seperti 1:1, 2:3, 0:0
        match_score = re.search(r'(\d+)\s*:\s*(\d+)', score_text)

        if match_score:
            skor = f"{match_score.group(1)}:{match_score.group(2)}"
        else:
            skor = None
    else:
        skor = None

    # -------------------------
    # VENUE
    # -------------------------
    td = row.find_all("td")

    if len(td) >= 6:
        venue = td[5].get_text(" ", strip=True)
    else:
        venue = None

    data_liga.append({
        "tanggal": tanggal,
        "tim_kandang": tim_kandang,
        "hasil_skor": skor,
        "tim_tandang": tim_tandang,
        "nama_venue": venue
    })

df_liga = pd.DataFrame(data_liga)

print("Ukuran:", df_liga.shape)
df_liga.head(35)
```

### 4.5 Scraping Liga 1 musim 2024/2025

```python
def scrape_liga1_persebaya(url, nama_event, id_prefix):

    response = requests.get(
        url,
        headers={'User-Agent': 'Mozilla/5.0'},
        timeout=30
    )

    response.raise_for_status()

    soup = BeautifulSoup(response.text, 'html.parser')

    # Bagian fixtures
    fixtures = soup.find("div", id="fixtures")

    if fixtures is None:
        print("Bagian fixtures tidak ditemukan.")
        return pd.DataFrame()

    # Semua tabel pertandingan
    tables = fixtures.find_all(
        "table",
        id="table-pertandingan"
    )

    # Semua container tanggal
    containers = fixtures.find_all(
        "div",
        class_="col-md-12"
    )

    # Container pertama = wrapper seluruh fixtures
    tanggal_containers = containers[1:]

    print("Jumlah pertandingan :", len(tables))
    print("Jumlah tanggal      :", len(tanggal_containers))

    # Cek kesesuaian
    if len(tables) != len(tanggal_containers):
        print("PERINGATAN: jumlah tabel dan tanggal berbeda.")

    # Nama bulan Indonesia
    bulan_id = {
        "Januari": 1,
        "Februari": 2,
        "Maret": 3,
        "April": 4,
        "Mei": 5,
        "Juni": 6,
        "Juli": 7,
        "Agustus": 8,
        "September": 9,
        "Oktober": 10,
        "November": 11,
        "Desember": 12
    }

    def parse_tanggal_id(tanggal):

        match = re.search(
            r'(\d{1,2})\s+'
            r'(Januari|Februari|Maret|April|Mei|Juni|Juli|'
            r'Agustus|September|Oktober|November|Desember)'
            r'\s+(\d{4})',
            tanggal,
            re.IGNORECASE
        )

        if not match:
            return pd.NaT

        hari = int(match.group(1))
        bulan = bulan_id[match.group(2).capitalize()]
        tahun = int(match.group(3))

        return pd.Timestamp(
            year=tahun,
            month=bulan,
            day=hari
        )

    data = []

    # Gunakan pasangan berdasarkan urutan
    for i, (table, container) in enumerate(
        zip(tables, tanggal_containers),
        start=1
    ):

        # -------------------------
        # TANGGAL
        # -------------------------
        tanggal_text = container.get_text(
            " ",
            strip=True
        )

        tanggal = parse_tanggal_id(tanggal_text)

        # -------------------------
        # BARIS PERTANDINGAN
        # -------------------------
        row = table.find(
            "tr",
            class_="table-row"
        )

        if row is None:
            continue

        # -------------------------
        # TIM
        # -------------------------
        tim = row.select(
            "b.club-fixture-team"
        )

        if len(tim) < 2:
            continue

        tim_kandang = tim[0].get_text(
            " ",
            strip=True
        )

        tim_tandang = tim[1].get_text(
            " ",
            strip=True
        )

        # -------------------------
        # SKOR
        # -------------------------
        score_element = row.select_one(
            ".club-fixture-score"
        )

        skor = None

        if score_element:

            score_text = score_element.get_text(
                " ",
                strip=True
            )

            match_score = re.search(
                r'(\d+)\s*:\s*(\d+)',
                score_text
            )

            if match_score:
                skor = (
                    f"{match_score.group(1)}:"
                    f"{match_score.group(2)}"
                )

        # -------------------------
        # VENUE
        # -------------------------
        td = row.find_all("td")

        venue = None

        if len(td) >= 6:
            venue = td[5].get_text(
                " ",
                strip=True
            )

        # -------------------------
        # SIMPAN
        # -------------------------
        data.append({
            "id_event": f"{id_prefix}-{i:03d}",
            "nama_event": nama_event,
            "jenis_olahraga": "Sepak Bola",
            "tanggal_mulai": tanggal,
            "tanggal_selesai": tanggal,
            "nama_venue": venue,
            "level_event": "Nasional",
            "peserta_tim": (
                f"{tim_kandang} vs "
                f"{tim_tandang}"
            ),
            "hasil_skor": skor,
            "sumber": url
        })

    df = pd.DataFrame(data)

    # Hanya pertandingan di Gelora Bung Tomo
    df = df[
        df["nama_venue"].str.contains(
            "GELORA BUNG TOMO",
            case=False,
            na=False
        )
    ].copy()

    # Reset index
    df = df.reset_index(drop=True)

    # ID ulang khusus event GBT
    df["id_event"] = [
        f"{id_prefix}-{i:03d}"
        for i in range(1, len(df) + 1)
    ]

    return df
```

### 5.5 Memisahkan nama peserta dan skor

```python
def pisahkan_skor(teks):
    """
    Mengubah:
    'Laos 0-6 Australia'
    menjadi:
    peserta = 'Laos vs Australia'
    skor = '0:6'
    """

    match = re.match(
        r'^(.+?)\s+(\d+)\s*[-:]\s*(\d+)\s+(.+)$',
        str(teks).strip()
    )

    if match:
        tim_home = match.group(1).strip()
        skor_home = match.group(2)
        skor_away = match.group(3)
        tim_away = match.group(4).strip()

        peserta = f"{tim_home} vs {tim_away}"
        skor = f"{skor_home}:{skor_away}"

        return peserta, skor

    return str(teks).strip(), None


hasil_pisah = df_aff['peserta_tim'].apply(pisahkan_skor)

df_aff['peserta_tim'] = hasil_pisah.apply(lambda x: x[0])

# Hanya isi skor kalau sebelumnya belum ada
df_aff['hasil_skor'] = df_aff.apply(
    lambda row: row['hasil_skor']
    if pd.notna(row['hasil_skor'])
    else hasil_pisah.loc[row.name][1],
    axis=1
)

display(
    df_aff[
        [
            'id_event',
            'tanggal_mulai',
            'nama_venue',
            'peserta_tim',
            'hasil_skor'
        ]
    ]
)
```

```python
import requests
import re
import pandas as pd
from bs4 import BeautifulSoup

# =========================================================
# 1. URL U-15
# =========================================================

url_u15 = "https://pssijatim.com/page-piala-soeratin-u-15"

headers = {
    "User-Agent": "Mozilla/5.0"
}


# =========================================================
# 2. REQUEST
# =========================================================

response_u15 = requests.get(
    url_u15,
    headers=headers,
    timeout=60
)

response_u15.raise_for_status()

soup_u15 = BeautifulSoup(
    response_u15.text,
    "html.parser"
)

rows_u15 = soup_u15.select(
    "tr.sp-row.sp-post"
)

print(
    f"U-15: ditemukan {len(rows_u15)} baris pertandingan"
)


# =========================================================
# 3. PEMETAAN BULAN INDONESIA
# =========================================================

bulan_indo = {
    "Januari": 1,
    "Februari": 2,
    "Maret": 3,
    "April": 4,
    "Mei": 5,
    "Juni": 6,
    "Juli": 7,
    "Agustus": 8,
    "September": 9,
    "Oktober": 10,
    "November": 11,
    "Desember": 12
}


# =========================================================
# 4. FUNGSI TANGGAL
# =========================================================

def parse_tanggal_u15(row):

    for a in row.find_all("a"):

        text = a.get_text(
            " ",
            strip=True
        )

        match = re.search(
            r'(\d{1,2})\s+'
            r'(Januari|Februari|Maret|April|Mei|Juni|'
            r'Juli|Agustus|September|Oktober|November|Desember)'
            r'\s+(\d{4})',
            text,
            re.IGNORECASE
        )

        if match:

            hari = int(match.group(1))

            nama_bulan = match.group(2).capitalize()

            tahun = int(match.group(3))

            nomor_bulan = bulan_indo[nama_bulan]

            return pd.Timestamp(
                year=tahun,
                month=nomor_bulan,
                day=hari
            )

    return None


# =========================================================
# 5. FUNGSI SKOR
# =========================================================

def parse_skor_u15(row):

    # Cari semua link dalam baris pertandingan
    semua_a = row.find_all("a")

    for a in semua_a:

        text = a.get_text(
            " ",
            strip=True
        )

        # Pola:
        # 3 - 2
        # 1 - 0
        # 2-1

        match = re.fullmatch(
            r'\s*(\d+)\s*-\s*(\d+)\s*',
            text
        )

        if match:

            return (
                f"{match.group(1)}:"
                f"{match.group(2)}"
            )

    return None


# =========================================================
# 6. FUNGSI LEVEL / BABAK
# =========================================================

def parse_level_u15(row):

    league_element = row.select_one(
        "div.sp-event-league"
    )

    if not league_element:
        return None

    league_text = league_element.get_text(
        " ",
        strip=True
    )

    # Contoh:
    # Grup H Piala Soeratin U-15
    # Final Juara 1 Piala Soeratin U-15
    # Final Juara 3 Piala Soeratin U-15

    league_text = re.sub(
        r'\s+Piala Soeratin U-15\s*$',
        '',
        league_text,
        flags=re.IGNORECASE
    )

    return league_text.strip()


# =========================================================
# 7. SCRAPING
# =========================================================

data_u15 = []

for i, row in enumerate(
    rows_u15,
    start=1
):

    # -----------------------------------------------------
    # TANGGAL
    # -----------------------------------------------------

    tanggal = parse_tanggal_u15(row)


    # -----------------------------------------------------
    # PESERTA / TIM
    # -----------------------------------------------------

    title = row.select_one(
        "h4.sp-event-title"
    )

    if title:

        peserta_tim = title.get_text(
            " ",
            strip=True
        )

    else:

        peserta_tim = None


    # -----------------------------------------------------
    # VENUE
    # -----------------------------------------------------

    semua_venue = row.select(
        "div.sp-event-venue"
    )

    if semua_venue:

        venue = semua_venue[0].get_text(
            " ",
            strip=True
        )

    else:

        venue = None


    # -----------------------------------------------------
    # LEVEL / BABAK
    # -----------------------------------------------------

    level_event = parse_level_u15(row)


    # -----------------------------------------------------
    # HASIL SKOR
    # -----------------------------------------------------

    hasil_skor = parse_skor_u15(row)


    # -----------------------------------------------------
    # SIMPAN
    # -----------------------------------------------------

    data_u15.append({

        "id_event":
            f"SOERATIN-U15-{i:03d}",

        "nama_event":
            "Piala Soeratin U-15",

        "jenis_olahraga":
            "Sepak Bola",

        "tanggal_mulai":
            tanggal,

        "tanggal_selesai":
            tanggal,

        "nama_venue":
            venue,

        "level_event":
            level_event,

        "peserta_tim":
            peserta_tim,

        "hasil_skor":
            hasil_skor,

        "sumber":
            url_u15
    })


# =========================================================
# 8. DATAFRAME
# =========================================================

kolom_event = [
    "id_event",
    "nama_event",
    "jenis_olahraga",
    "tanggal_mulai",
    "tanggal_selesai",
    "nama_venue",
    "level_event",
    "peserta_tim",
    "hasil_skor",
    "sumber"
]

df_u15 = pd.DataFrame(
    data_u15
)[kolom_event]


# =========================================================
# 9. HASIL
# =========================================================

print(
    "Ukuran data U-15:",
    df_u15.shape
)

display(
    df_u15.head()
)
```
